```python
import pandas as pd
import numpy as np

mainData = pd.read_csv('Data/US-pumpkins.csv')
```

```python
columns_selected = ['City Name', 'Package', 'Variety', 'Origin', 'Item Size','Color']
mainData = mainData[columns_selected]
mainData.dropna(inplace=True)
mainData.head(50)
```

```python
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder
from sklearn.compose import ColumnTransformer

#Ordinal Encoding
item_size_categories = [['sml', 'med', 'med-lge', 'lge', 'xlge', 'jbo', 'exjbo']]
ordinal_feature = ['Item Size']
ordinal_encoder =  OrdinalEncoder(categories=item_size_categories)

#OneHot Encoding
onehot_feature = ['City Name', 'Package', 'Variety', 'Origin']
onehot_encoder = OneHotEncoder(sparse_output=False)

#Compose columns

ct = ColumnTransformer(transformers=[
    ('ord', ordinal_encoder, ordinal_feature),
    ('cat', onehot_encoder, onehot_feature)
])
ct.set_output(transform='pandas')
feature_encoded = ct.fit_transform(mainData)

#Label Encoding
label_encoder = LabelEncoder()
label_encoded = label_encoder.fit_transform(mainData['Color'])

#Merging
newData = feature_encoded.assign(Target_color=label_encoded)
newData.head(50)
```

```python
from sklearn import model_selection, linear_model
from sklearn.metrics import f1_score, classification_report, confusion_matrix

X = newData[newData.columns.difference(['Target_color'])]
y = newData['Target_color']

X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.2, random_state=0)
model = linear_model.LogisticRegression()
model.fit(X_train, y_train)
y_hat = model.predict(X_test)

print(classification_report(y_test, y_hat))
print(f"f1-score: {f1_score(y_test, y_hat):.3f} \n")
print(confusion_matrix(y_test, y_hat))
```
